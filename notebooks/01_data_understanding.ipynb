# 01 Data Understanding

## Melbourne Metro Rail Patronage Analytics Project

This notebook explores the structure, quality, and consistency of the raw Melbourne metropolitan train station patronage datasets before data cleaning and consolidation.

## 1. Objective

The purpose of this notebook is to:

- Load and inspect the raw patronage datasets
- Understand available columns and data types
- Review row and column counts
- Check missing values and duplicates
- Compare file structures across years
- Identify early data quality issues before creating the master dataset

```python
# Import Libraries
import pandas as pd
from pathlib import Path
```

```python
# Set Project Paths
PROJECT_ROOT = Path("..")

RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DATA_DIR = PROJECT_ROOT / "data" / "processed"

RAW_DATA_DIR
```

```python
# List Available Raw Files
raw_files = sorted(RAW_DATA_DIR.glob("*.csv"))

raw_files
```

```python
# Create File Inventory
file_inventory = []

for file in raw_files:
    df_temp = pd.read_csv(file)
    
    file_inventory.append({
        "file_name": file.name,
        "rows": df_temp.shape[0],
        "columns": df_temp.shape[1],
        "column_names": list(df_temp.columns)
    })

inventory_df = pd.DataFrame(file_inventory)

inventory_df[["file_name", "rows", "columns"]]
```

```python
# Load Latest Dataset
latest_file = RAW_DATA_DIR / "annual_metropolitan_train_station_entries_fy_2024_2025.csv"

df = pd.read_csv(latest_file)

df.head()
```

```python
# Check shape
df.shape
```

The latest dataset contains 222 rows and 16 columns.

```python
# Inspect Column Names
df.columns.tolist()
```

```python
# Inspect Data Types
df.info()
```

```python
# Summary Statistics
df.describe(include="all")
```

```python
# Missing Value Check
missing_values = df.isna().sum().reset_index()
missing_values.columns = ["column", "missing_count"]

missing_values
```

```python
# Duplicate Check
duplicate_count = df.duplicated().sum()

duplicate_count
```

```python
# Compare Column Consistency Across Files
reference_columns = set(pd.read_csv(raw_files[0]).columns)

column_check = []

for file in raw_files:
    temp_df = pd.read_csv(file)
    current_columns = set(temp_df.columns)
    
    column_check.append({
        "file_name": file.name,
        "columns_match_reference": current_columns == reference_columns,
        "missing_columns": list(reference_columns - current_columns),
        "extra_columns": list(current_columns - reference_columns)
    })

column_check_df = pd.DataFrame(column_check)

column_check_df
```

```python
# Row Count Difference
inventory_df[["file_name", "rows", "columns"]].sort_values("rows", ascending=False)
```
